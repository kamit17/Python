```python
def filter_solutions():
    with open("solutions.csv","r") as source_file, open("correct.csv","w") as correct_file, open("incorrect.csv","w") as incorrect_file:
        for line in source_file:
            parts = line.strip().split(";")
            name = parts[0]
            problem = parts[1]
            answer = parts[2]
            if "+" in problem:
                numbers = problem.split("+")
                correct_result = int(numbers[0]) + int(numbers[1])
            else:
                numbers = problem.split("-")
                correct_result = int(numbers[0]) - int(numbers[1])
            if correct_result ==int(answer):
                correct_file.write(line)
            else:
                incorrect_file.write(line)
            
        #print(contents)
filter_solutions()
```
