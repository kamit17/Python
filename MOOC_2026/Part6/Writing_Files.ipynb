```python
def filter_solutions():
    with open("solutions.csv","r") as source_file, open("correct.csv","w") as correct_file, open("incorrect.csv","w") as incorrect_file:
        for line in source_file:
            parts = line.strip().split(";")
            name = parts[0]
            problem = parts[1]
            answer = parts[2]
            if "+" in problem:
                numbers = problem.split("+")
                correct_result = int(numbers[0]) + int(numbers[1])
            else:
                numbers = problem.split("-")
                correct_result = int(numbers[0]) - int(numbers[1])
            if correct_result ==int(answer):
                correct_file.write(line)
            else:
                incorrect_file.write(line)
            
        #print(contents)
filter_solutions()
```

```python
# read a csv file
#calculate the points total and determines the grade
#writing the file

#read file -- store data in a dictionary

def read_weekly_points(filename):
    weekly_points = {}
    with open(filename) as my_file:
        for line in my_file:
            parts = line.split(";")
            points_list = []
            for points in parts[1:]:
                points_list.append(int(points))
            weekly_points[parts[0]] = points_list
    return weekly_points
                    
#determining the grade based on the points received
def grade(points):
    if points < 20:
        return 0
    elif points < 25:
        return 1
    elif points < 30:
        return 2
    elif points < 35:
        return 3
    elif points < 40:
        return 4
    else:
        return 5
                    
#writing the results to the file
def save_results(filename, weekly_points):
    with open(filename, "w") as my_file:
        for name, points_list in weekly_points.items():
            point_sum = sum(points_list)
            my_file.write(f"{name};{point_sum};{grade(point_sum)}\n")

#print out grades for a single student
def get_grade(student_name,weekly_points):
    for name, points_list in weekly_points.items():
        if name == student_name:
            return grade(sum(points_list))
            
weekly_points = read_weekly_points("weekly_points.csv")
save_results("results.csv", weekly_points)
print(get_grade("Paula",weekly_points))
```

```python
def find_words(search_term: str):
    words = []

    with open("words.txt", "r") as new_file:
        for line in new_file:
            word = line.strip()

            # * at the end
            if search_term.endswith("*"):
                start = search_term[:-1]

                if word.startswith(start):
                    words.append(word)

            # * at the beginning
            elif search_term.startswith("*"):
                end = search_term[1:]

                if word.endswith(end):
                    words.append(word)

            # . wildcard
            elif "." in search_term:

                if len(word) == len(search_term):
#Assume it is True .
                    match = True

                    for i in range(len(word)):

                        if search_term[i] != "." and search_term[i] != word[i]:
                            match = False

                    if match:
                        words.append(word)

            # exact match
            else:

                if word == search_term:
                    words.append(word)

    return words

print(find_words("can"))     # exact match
print(find_words("ca*"))     # starts with ca
print(find_words("*ane"))    # ends with ane
print(find_words("p.ng"))    # ping, pong, etc.
```
