```python
#Please write a function named largest, which reads the file and returns the largest number in the file.
def largest():
    numbers = []
    with open("numbers.txt") as new_file:
        for line in new_file:
            numbers.append(int(line))

        if len(numbers) > 0:
            largest_num = numbers[0]
            for num in numbers:
                if num > largest_num:
                    largest_num = num
            return largest_num

print("Largest number is : ",largest())  
```

```python
def largest():
    largest_num = None
    with open("numbers.txt") as new_file:
        for line in new_file:
            num = int(line.strip())
            if largest_num is None or num > largest_num:
                largest_num = num
    return largest_num

print("Largest number is : ", largest())
```

```python
def largest():
    numbers = []
    with open("numbers.txt") as new_file:
        for line in new_file:
            numbers.append(int(line.strip()))  # strip removes newline

    if numbers:  # make sure the list isn’t empty
        return max(numbers)   # built-in max does the job
    else:
        return None

print("Largest number is : ", largest())
```

```python
def read_fruits():
    fruits = {}
    with open("fruits.csv") as new_file:
        for line in new_file:
            line = line.strip()   #remove new line
            name,price = line.split(";")  #split into two parts
            fruits[name] = float(price) #store in dictionary
    return fruits

print(read_fruits())

```

```python
with open("matrix.txt") as new_file:
    matrix= []
    for line in new_file:
        #remove newline at the end
        line = line.strip()
        #split by commas into a list of string
        parts = line.split(",")
        #convert each string into an integer
        row = [int(x) for x in parts]
        #add the row to the matrix
        matrix.append(row)
print(matrix)

def matrix_sum():
    matrix = []
    with open("matrix.txt") as new_file:
        for line in new_file:
            row = [int(x) for x in line.strip().split(",")]
            matrix.append(row)
    return sum(sum(row) for row in matrix)
#
print(matrix_sum())

def matrix_max():
    matrix = []
    with open("matrix.txt") as new_file:
        for line in new_file:
            row = [int(x) for x in line.strip().split(",")]
            matrix.append(row)
    return max(max(row) for row in matrix)
print(matrix_max())

def row_sums():
    matrix = []
    with open("matrix.txt") as new_file:
        for line in new_file:
            row = [int(x) for x in line.strip().split(",")]
            matrix.append(row)

    return [sum(row) for row in matrix]

print(row_sums())    
```
