```python
def create_tuple(x: int, y: int, z: int):
    #first element is the smallest of the arguments
    first = min(x,y,z)
    #second argument is the greatest 
    second = max(x,y,z)
    #third element in the tuple is the sum of the arguments
    third = x+y+z

    return(first,second,third)

print(create_tuple(5, 3, -1))
```

```python
def oldest_person(people):
    #set the oldest year and name to the first tuple year
    oldest_year = people[0][1]
    oldest_name = people[0][0]

    for item in people:
        name = item[0]
        year = item[1]

        if year < oldest_year:
            oldest_year = year
            oldest_name = name

    return oldest_name

p1 = ("Adam", 1977)
p2 = ("Ellen", 1985)
p3 = ("Mary", 1953)
p4 = ("Ernest", 1997)
people = [p1, p2, p3, p4]

print(oldest_person(people))
```

```python
def older_people(people, year):
    older = []

    for person in people:
        name = person[0]
        year_born = person[1]

        if year_born < year:
            older.append(name)

    return older
    
p1 = ("Adam", 1977)
p2 = ("Ellen", 1985)
p3 = ("Mary", 1953)
p4 = ("Ernest", 1997)
people = [p1, p2, p3, p4]

older = older_people(people, 1979)
print(older)
```
