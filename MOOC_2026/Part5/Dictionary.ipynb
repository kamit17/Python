```python

def times_ten(start_index,end_index):
    result = {}  #initialize empty dicitonary
    for key in range(start_index, end_index + 1):
        result[key] = key * 10  #dict[key] = value
    return result
d = times_ten(3,6)
print(d)
```

```python
def factorials(n):
    result = {}

    for number in range(1, n + 1):

        factorial = 1

        for i in range(1, number + 1):
            factorial *= i

        result[number] = factorial

    return result

k = factorials(5)
print(k[1])
print(k[3])
print(k[5])
```

```python
word_list = [
  "banana", "milk", "beer", "cheese", "sourmilk", "juice", "sausage",
  "tomato", "cucumber", "butter", "margarine", "cheese", "sausage",
  "beer", "sourmilk", "sourmilk", "butter", "beer", "chocolate"
]

#how many times each word appears in the list.
def counts(my_list):
    words = {}
    for word in my_list:
        #if the word is not yet in the dcitionary,initialize the value to zero
        if word not in words:
            words[word] = 0
        #increment the value
        words[word] += 1
    return words
print(counts(word_list))
        
```

```python
#categorize words based on the intial letter in each word

def categorize_by_initial(my_list):
    groups = {}
    for word in my_list:
        initial = word[0]
        #initialize a new list when the letter is first encountered
        if initial not in groups:
            groups[initial] = []
        #add the word to the appropriate list
        groups[initial].append(word)
    return groups

groups = categorize_by_initial(word_list)

for key,value in groups.items():
    print(f"words beginning with {key}:")
    for word in value:
        print(word)
```

```python
def histogram(input_string):
    words = {}
    for character in input_string:
        if character not in words:
            words[character] = 0
        words[character] += 1
    for letter,count in words.items():
        print (letter, "*" * count)

histogram("statistically")   
```

Please write a function named invert(dictionary: dict), which takes a dictionary as its argument. The dictionary should be inverted in place so that values become keys and keys become values.

An example of its use:

s = {1: "first", 2: "second", 3: "third", 4: "fourth"}
invert(s)
print(s)
Sample output
{"first": 1, "second": 2, "third": 3, "fourth": 4}

```python
def invert(dictionary: dict):
    new_dict = {}
    for key,value in dictionary.items():
        new_dict[value] = key
#print(s1)
    dictionary.clear()
    dictionary.update(new_dict)

s = {1: "first", 2: "second", 3: "third", 4: "fourth"}
invert(s)
print(s)
```
