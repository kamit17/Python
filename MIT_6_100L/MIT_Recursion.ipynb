```python
#functin that calculates n^p for variables n and p
def power_recur(n,p):
    if p == 0:
        return 1
    elif p == 1: 
        return n
    else:
        return n*power_recur(n,p-1)
print(power_recur(2,3))
```

```python
#factorial
#n! = n*(n-1) * (n -2 ) * (n-1) * ...* 1

def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```
